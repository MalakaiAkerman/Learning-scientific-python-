```python
#Defining a single random event function
def coin_flip(p):
  from random import random
  n = random()
  if n<p:
    return 'Heads'
  else:
    return 'Tails'
coin_flip(0.5)
```

```python
#function for repeating multiple random events
def multiple_flips(n,p):
  head_count = 0
  for i in range(n):
    result = coin_flip(p)
    if result != 'Heads':
      continue
    head_count+=1
  head_fraction = head_count/n
  return head_fraction
multiple_flips(100,0.5)
```

```python
#plotting graph of
def multiple_flips_graph(p,n):
  import numpy as np
  import matplotlib.pyplot as plt

  x = np.arange(1,n+1)
  y = []
  head_count = 0

  for num in x:
    result = coin_flip(p)
    if result == 'Heads':
      head_count+=1
    else:
      pass
    y.append(head_count/num)
  plt.title('Graph of simulation probability tending towards actual probability over a large number of coin flips')
  plt.xlabel('number of flips')
  plt.ylabel('simulated probability')
  plt.ylim(0,1)
  plt.plot(x,y)
  plt.axhline(p, color='red', linestyle='--')

multiple_flips_graph(0.67,10000)
```
