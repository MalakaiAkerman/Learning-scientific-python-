Small project practising calculating failure probabilities with or without redundancy of components

```python
#simplest case, single component
def fail_p_single():
  p = float(input())
  if 0<=p<=1:
    print(p)
  else:
    print('enter a valid probability')
fail_p_single()
```

```python
#two components, both must work
def fail_p_double():
  p = float(input())
  if 0<=p<=1:
    return 1-(1-p)**2
  else:
    print('enter a valid probability')
fail_p_double()
```

```python
#two components, redundancy
def fail_p_double_redundant():
  p = float(input())
  if 0<=p<=1:
    prob = round(p*p,4)
    return prob
  else:
    print('enter a valid probability')
fail_p_double_redundant()
```
